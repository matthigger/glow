```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Path where your CSVs are stored
folder = '/home/matt/.local/share/hglm/results/exp_25-08-14_1551/out'
folder_path = Path(folder)

# Find all CSV files that match the pattern
csv_files = list(folder_path.glob("*_alpha_tailor.csv"))

# Read and combine into a single DataFrame
dfs = [pd.read_csv(file) for file in csv_files]
all_df = pd.concat(dfs, ignore_index=True)
```

```python
metric = 'f1'
avg_df = (
    all_df.groupby('alpha_tailor')[metric]
    .mean()
    .reset_index()
    .sort_values(by='alpha_tailor')
)
```

```python

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharex=True)

metrics = ['f1', 'sens', 'spec']
for ax, metric in zip(axes, metrics):
    # Thin lines for each seed — sorted
    for seed, group in all_df.groupby('seed'):
        group_sorted = group.sort_values(by='alpha_tailor')
        ax.plot(group_sorted['alpha_tailor'], group_sorted[metric],
                color='gray', alpha=0.5, linewidth=0.8)
    
    # Bold average line — also sorted
    avg_df = (
        all_df.groupby('alpha_tailor')[metric]
        .mean()
        .reset_index()
        .sort_values(by='alpha_tailor')
    )
    ax.plot(avg_df['alpha_tailor'], avg_df[metric],
            color='red', linewidth=2, label='Average')
    
    ax.set_xlabel('alpha_tailor')
    ax.set_ylabel(metric)
    ax.set_title(metric)
    ax.legend()

plt.tight_layout()
plt.show()
```
