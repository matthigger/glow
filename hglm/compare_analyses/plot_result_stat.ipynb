```python
from hglm.compare_analyses import load_update_all, extract, plot_feats

folder = '/home/matt/Dropbox/pnl_hglm/results/exp_25Aug06-1047/'
df = load_update_all(folder)
f_list, seed_list, score_dict = extract(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Optional: load your DataFrame ---
# df = pd.read_csv('your_file.csv')

# Define metrics to plot
metrics = ['f1', 'sens', 'spec']

# Create color mapping per stat
palette = sns.color_palette('tab10', n_colors=df['stat'].nunique())
stat_colors = dict(zip(df['stat'].unique(), palette))

# Create 1x3 subplot grid
fig, axes = plt.subplots(1, 3, figsize=(12, 4), sharex=True)

# Iterate over metrics and axes
for ax, metric in zip(axes, metrics):
    
    # Plot thin lines: (stat, seed) — NO legend
    for (stat, seed), group in df.groupby(['stat', 'seed']):
        group = group.sort_values('f_ratio')
        ax.plot(
            group['f_ratio'], group[metric],
            color=stat_colors[stat],
            alpha=0.4,
            linewidth=1,
            label=None
        )
    
    # Plot bold lines: average per stat — LEGEND
    for stat, group in df.groupby('stat'):
        avg_group = group.groupby('f_ratio', as_index=False).mean(numeric_only=True)
        avg_group = avg_group.sort_values('f_ratio')
        ax.plot(
            avg_group['f_ratio'], avg_group[metric],
            color=stat_colors[stat],
            linewidth=3,
            label=f"{stat} (avg)"
        )
    
    # Configure axis
    ax.set_xscale('log')
    ax.set_xlabel('f_ratio')
    ax.set_ylabel(metric)
    ax.set_title(f'{metric.upper()} vs F-ratio')

# Place legend only once (on the last axis)
axes[-1].legend(title='Stat', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='small')

plt.tight_layout()
plt.show()
```
