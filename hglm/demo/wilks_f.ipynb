```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def simulate_wilks_lambda(n=100, p=3, m=2, num_simulations=10000):
    """
    Simulate Wilks' Lambda and compare empirical F-distribution to theoretical F-distribution.
    """
    # Generate fixed predictor matrix X (n x p)
    X = np.random.randn(n, p)
    
    # Generate true coefficients B (p x m)
    B = np.random.randn(p, m)
    
    # Projection matrix onto column space of X
    P_X = X @ np.linalg.pinv(X)
    
    # Identity matrix
    I_n = np.eye(n)
    
    # Corrected degrees of freedom for F-distribution transformation
    d1 = m * (p - 1)
    d2 = (n - p - m + 1) - (m - 1)
    
    # Storage for F-statistics
    f_values = []
    
    for _ in range(num_simulations):
        # Generate random error matrix E (n x m)
        E = np.random.randn(n, m)
        Y = E
        
        # Generate response matrix Y (n x m)
        # Y = X @ B + E
        
        # Compute Hypothesis Sum of Squares and Cross-Products (H)
        H = Y.T @ P_X @ Y  # Explained variance
        
        # Compute Error Sum of Squares and Cross-Products (E)
        E_matrix = Y.T @ (I_n - P_X) @ Y  # Residual variance
        
        # Compute Wilks' Lambda using log-determinants for numerical stability
        log_Lambda = np.linalg.slogdet(E_matrix)[1] - np.linalg.slogdet(H + E_matrix)[1]
        Lambda = np.exp(log_Lambda)
        
        # Transform Lambda into an F-statistic
        s = min(m, p - 1)
        F_stat = ((1 - Lambda**(1/s)) / (Lambda**(1/s))) * (d2 / d1)
        f_values.append(F_stat)
    
    # Plot histogram of empirical F-statistics
    plt.figure(figsize=(8, 5))
    plt.hist(f_values, bins=50, density=True, alpha=0.6, color='b', label='Empirical F')
    
    # Overlay theoretical F-distribution
    x = np.linspace(0, max(f_values), 1000)
    plt.plot(x, stats.f.pdf(x, d1, d2), 'r-', lw=2, label=f'Theoretical F({d1},{d2})')
    
    plt.xlabel('F-statistic')
    plt.ylabel('Density')
    plt.title('Empirical vs Theoretical F-distribution from Wilks’ Lambda')
    plt.legend()
    plt.show()

# Run the simulation
simulate_wilks_lambda()
```
